```python
# 1. Median horsepower
original_median = (df["horsepower"].median())

# 2. Most frequent horsepower
most_frequent = df["horsepower"].mode()[0]

# 3. Fill missing horsepower values
df["horsepower"] = df["horsepower"].fillna(most_frequent)

# 4. Median horsepower after filling missing values
new_median = df["horsepower"].median()

print(original_median)
print(new_median)
```

```python
# select all cars from Asia
asia = df[df["origin"] == "Asia"]

# select vehicle_weight and model_year
asia_selected = asia[["vehicle_weight", "model_year"]]

# select the first 7 values
asia_first7 = asia_selected.head(7)

# get the underlying NumPy array
X = asia_first7.values

# compute XTX
XTX = np.dot(X.T, X)

# invert XTX
XTX_inv = np.linalg.inv(XTX)

# create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# calculate w
w = np.dot(np.dot(XTX_inv, X.T), y)

# sum all elements of w
w.sum()
```
